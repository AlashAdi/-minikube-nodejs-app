```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score
from sklearn.decomposition import PCA

# ==========================================
# 1. Деректерді жүктеу және StandardScaler масштабтау
# ==========================================
wine = load_wine()
X = wine.data
print("Wine деректерінің өлшемі (X.shape):", X.shape)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Метрикаларды есептейтін көмекші функция
def get_metrics(X_data, labels):
    mask = labels != -1
    if len(np.unique(labels[mask])) <= 1:
        return None, None, None
    sil = silhouette_score(X_data[mask], labels[mask])
    db = davies_bouldin_score(X_data[mask], labels[mask])
    ch = calinski_harabasz_score(X_data[mask], labels[mask])
    return sil, db, ch

# ==========================================
# 2. K-means: k = 2...7 үшін Elbow және Silhouette
# ==========================================
inertias = []
sil_scores = []
k_range = range(2, 8)

for k in k_range:
    km = KMeans(n_clusters=k, n_init=10, random_state=42)
    labels = km.fit_predict(X_scaled)
    inertias.append(km.inertia_)
    sil_scores.append(silhouette_score(X_scaled, labels))

# Elbow және Silhouette графиктер
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(list(k_range), inertias, marker='o', color='blue')
plt.title("Wine: Elbow әдісі")
plt.xlabel("k кластер саны")
plt.ylabel("Inertia")
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(list(k_range), sil_scores, marker='o', color='green')
plt.title("Wine: Silhouette әдісі")
plt.xlabel("k кластер саны")
plt.ylabel("Silhouette Score")
plt.grid(True)
plt.show()

# ==========================================
# 3. Модельдерді есептеу
# ==========================================
best_k = 3  # Elbow және Silhouette нәтижесі бойынша

# K-means
kmeans = KMeans(n_clusters=best_k, n_init=10, random_state=42)
labels_km = kmeans.fit_predict(X_scaled)

# Agglomerative (ward)
agg = AgglomerativeClustering(n_clusters=best_k, linkage='ward')
labels_agg = agg.fit_predict(X_scaled)

# DBSCAN параметрін таңдау (eps = 0.5, 0.7, 0.9, 1.1, 1.3; min_samples = 5)
eps_values = [0.5, 0.7, 0.9, 1.1, 1.3]
db_results = []

for eps in eps_values:
    db = DBSCAN(eps=eps, min_samples=5)
    lbls = db.fit_predict(X_scaled)
    n_noise = np.sum(lbls == -1)
    n_clusters = len(set(lbls)) - (1 if -1 in lbls else 0)
    sil, db_idx, ch_idx = get_metrics(X_scaled, lbls)
    db_results.append([eps, n_clusters, n_noise, sil, db_idx, ch_idx])

db_df = pd.DataFrame(db_results, columns=['eps', 'clusters', 'noise', 'silhouette', 'davies_bouldin', 'calinski_harabasz'])
print("\n=== DBSCAN параметрлерін зерттеу кестесі ===")
print(db_df.to_string(index=False))

# Негізгі DBSCAN модель (eps = 1.1, min_samples = 5)
db_best = DBSCAN(eps=1.1, min_samples=5)
labels_db = db_best.fit_predict(X_scaled)

# ==========================================
# 4. PCA 2D визуализация
# ==========================================
pca = PCA(n_components=2)
X_2d = pca.fit_transform(X_scaled)

plt.figure(figsize=(15, 4))
plt.subplot(1, 3, 1)
plt.scatter(X_2d[:, 0], X_2d[:, 1], c=labels_km, cmap='viridis')
plt.title("K-means (k=3)")

plt.subplot(1, 3, 2)
plt.scatter(X_2d[:, 0], X_2d[:, 1], c=labels_agg, cmap='viridis')
plt.title("Agglomerative (ward, k=3)")

plt.subplot(1, 3, 3)
plt.scatter(X_2d[:, 0], X_2d[:, 1], c=labels_db, cmap='viridis')
plt.title("DBSCAN (eps=1.1, min_samples=5)")
plt.show()

# ==========================================
# 5. Салыстырмалы метрикалар кестесі
# ==========================================
models = {
    "K-means (k=3)": labels_km,
    "Agglomerative (ward, k=3)": labels_agg,
    "DBSCAN (eps=1.1)": labels_db
}

summary = []
for name, lbls in models.items():
    sil, db_i, ch_i = get_metrics(X_scaled, lbls)
    summary.append({
        "Algorithm": name,
        "Silhouette": round(sil, 4) if sil else None,
        "Davies-Bouldin": round(db_i, 4) if db_i else None,
        "Calinski-Harabasz": round(ch_i, 4) if ch_i else None
    })

summary_df = pd.DataFrame(summary)
print("\n=== Үш алгоритмнің салыстырмалы метрикалары ===")
print(summary_df.to_string(index=False))
```
